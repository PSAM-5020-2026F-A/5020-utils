### Simplify/Collapse/Fix Ingredients

```python
ings_fixed = {}

for r in recs:
  for i in r["ingredients"]:
    i = simplify(i)
    i = fix_map.get(i, i).strip()
    if len(i) > 0:
      ings_fixed[i] = ings_fixed.get(i, 0) + 1

print(len(ings_fixed))
ings_tuples = sorted(list(ings_fixed.items()), key=lambda x: x[1], reverse=True)

# unique ingredients
ings_unique_tuples = [i for i in ings_tuples if i[1] < 2]
len(ings_unique_tuples)
```

### Recipe vs Ingredient Analysis

```python
UNIQUE_THOLD = 4
tholds = range(2, 17, 2)
min_rec_pcts = []
recs_per_ings = []

for thold in tholds:
  recs_fixed = []
  recs_unique = []

  for r in recs:
    fixed_ings = set()
    has_unique = False
    for i in r["ingredients"]:
      i = simplify(i)
      i = fix_map.get(i, i).strip()
      # TODO: plurals -> singular
      if len(i) > 0:
        fixed_ings.add(i)
        has_unique = has_unique or (ings_fixed[i] <= thold)

    fixed_r = {
      "id": r["id"],
      "ingredients": list(fixed_ings)
    }
    if has_unique:
      recs_unique.append(fixed_r)
    else:
      recs_fixed.append(fixed_r)

  min_rec_pcts.append((thold+1) / len(recs_fixed))
  recs_per_ings.append(len(recs_fixed) / len(ings_fixed))

plt.plot(tholds, min_rec_pcts)
plt.title("Minimum Recipe %")
plt.show()

plt.plot(tholds, recs_per_ings)
plt.title("Recipe per Ingredient")
plt.show()
```

### Process Recipes

```python
MIN_REC_COUNT = 16
MIN_ING_COUNT = 5
MAX_ING_COUNT = 30
recs_fixed = []
recs_short = []
recs_unique = []

for r in recs:
  fixed_ings = set()
  has_unique = False
  for i in r["ingredients"]:
    i = simplify(i)
    i = fix_map.get(i, i).strip()
    if len(i) > 0:
      fixed_ings.add(i)
      has_unique = has_unique or (ings_fixed[i] < MIN_REC_COUNT)

  fixed_r = {
    "id": r["id"],
    "ingredients": list(fixed_ings)
  }

  if has_unique:
    recs_unique.append(fixed_r)
  elif (len(fixed_r["ingredients"]) < MIN_ING_COUNT) or (len(fixed_r["ingredients"]) > MAX_ING_COUNT):
    recs_short.append(fixed_r)
  else:
    recs_fixed.append(fixed_r)

print(f"Recipes: {len(recs_fixed)}")
print(f"Ingredients: {len(ings_fixed)}")
print(f"Recipes per Ingredient: {len(recs_fixed) / len(ings_fixed)}")
print(f"Min Ingredient per Recipe (%): {MIN_REC_COUNT / len(recs_fixed)}")

print(len(recs_fixed), len(recs_short), len(recs_unique))
```

### GenSim

```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
```

```python
docs = [TaggedDocument(doc, [i]) for i, doc in enumerate(recipes_cbow_df["recipe"].values)]
d2v = Doc2Vec(vector_size=8, window=2, min_count=5, workers=4)
d2v.build_vocab(docs)
d2v.train(docs, total_examples=d2v.corpus_count, epochs=40)
```
